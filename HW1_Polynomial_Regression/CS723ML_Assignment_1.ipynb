```python
X = np.asarray(features9)
target = np.asarray(t).reshape(-1)
print("Features:", X.shape)
print("Targets:", target.shape)
w = np.zeros(X.shape[1])
learning_rate = 0.001
iterations = 20000
lam = 0.01
print("Weights:", w.shape)
```

```python
loss_history = []

for epoch in range(iterations):
    predictions_reg = X @ w
    residuals = predictions_reg - target
    gradient = X.T @ residuals + lam * w
    w = w - learning_rate * gradient

    if epoch % 1000 == 0:
        loss = 0.5 * np.sum(residuals ** 2)
        penalty = 0.5 * lam * np.sum(w ** 2)
        loss_history.append(loss + penalty)

grid_reg = np.linspace(0, 1, 200).reshape(-1, 1)
grid_features_reg = poly9.transform(grid_reg)

print("Grid features:", grid_features_reg.shape)
print("Learned weights:", w.shape)


   
    



print("Final weights:", w)
print("Loss history:", loss_history)
```
